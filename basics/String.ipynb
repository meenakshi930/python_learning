# Modify Strings

```python
a = "Hello, World!"
print(a.upper())
a = "Hello, World!"
print(a.lower())
# The strip() method removes any whitespace from the beginning or the end:
a = " Hello, World! " 
print(a.strip()) 
# The replace() method replaces a string with another string:
a = "Hello, World!"
print(a.replace("H", "J"))
# The split() method returns a list where the text between the specified separator becomes the list items.
a = "Hello, World!"
print(a.split(",")) 
```

# 🔤 Case conversion

```python
# capitalize():First character uppercase, rest lowercase
a="hello WORLD".capitalize()
print(a)
# casefold():Strong lowercase conversion, useful for comparisons
b="HELLO"
print(b.casefold())
# swapcase():Upper ↔ lower
c="heLlo".swapcase()
print(c)
# title():First letter of every word uppercase
d="hello world".title()
print(d)
# lower():Convert to lowercase
e="HELLO".lower()
print(e)
# upper():Convert to uppercase
f="hello".upper()
print(f)
```

# 🔍 Searching & counting

```python
# count(): count occurrences
print("banana".count("n"))
# find: Find first occurrence; returns -1 if absent
print("hello".find("l"))
# rfind: find last occurence
print("hello".rfind("l"))
# index: find first occurence:give error if absent
print("hello".index("l")) 
# rindex(): Find last occurrence; gives error if absent
print("hello".rindex("l"))
# startswith(): check beginning
print("python".startswith("py"))
# endswith(): Check ending
print("python".endswith("on"))
```

# ✅ Checking string contents

```python
# isalnum: only letter/number
print("abc123".isalnum())
# isalpha: only letter
print("acv".isalpha())
# isascii: only ASCII characters
print("hello".isascii())
# isdecimal(): decimal character
print("123".isdecimal())
# isdigit: only digits
print("123".isdigit())
print("IV".isdigit())
# isnumeric: numeric characters(it accept roman)
print("123".isnumeric())
# islower: All applicable letters lowercase
print("hello".islower())
# isuper: All applicable letters uppercase
print("HELLO".isupper())
# isspace: only whitespace
print(" ".isspace())
# isprintable: All chahracter printable
print("hello".isprintable())
# isidentifier(): Valid Python variable name
print("ny_var".isidentifier())
```

# ✂️ Removing spaces

```python
# strip(): remove spaces from both ends
print("  hello world".strip())

# lstrip: remove from left
print(" hello".lstrip())

# rstrip(): remove from right
print("hello ".rstrip())
```

# 🔄 Splitting strings

```python
# split(): string --> list 
s="apple banana mango"
words=s.split()
print(words)
print("a.b.c".split("."))
# rsplit() : split from right
print("a-b-c".rsplit("-",1))
print("a-b-c".rsplit("-",2))
#splitlines(): Split at line breaks
print("d\ne\nf".splitlines())
```

```python
# joining string : join() does the opposite of split().
words = ["I", "love", "Python"]

result = " ".join(words)

print(result)
# I love Python
```

# 🔁 Replacing & translating

```python
# replace(): replace text
s="hello"
print(s.replace("l","m"))
# maketrans(): Create translation table(Used with translate())
# translate(): Replace multiple characters at once
table=str.maketrans("abc","123")
print("abcabc".translate(table))
```

```python
# format() is used to insert values into a string.
name="mee"
age=20
print("My name is {} and i am {} years old".format(name,age))

# format_map() does similar formatting using a dictionary:
data = {
    "name": "Mee",
    "age": 20
}

print("My name is {name} and I am {age}".format_map(data))
```
